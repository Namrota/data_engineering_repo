#### 2. Calculate session counts


Session count is a record of the number of times a user visits your website within a given timeframe, such as day, week, or month. A cookie on the visitor’s browser is used to track the user’s activity between the time they enter your domain and the time they leave.

Sessions create a record of all the interactions each visitor has on your website, including all the pages they view and the engagements they make with events, calls-to-action, forms, etc.


Assumption: 

- Every customer's first event starts a new session. Any later event after more than 30 minutes of inactivity starts another.
- NaT for a new session of a new customer when calculating the time difference, since no previous records to compare it with.

```python
session_df= valid_clickstream[['customer_id', 'event_timestamp']]
session_df.head(12)
```

```python
session_df= session_df.sort_values(['customer_id', 'event_timestamp'])
```

```python
session_df['time_diff'] = (
    session_df.groupby('customer_id')['event_timestamp']
    .diff()
)

session_df
```

```python
session_df['new_session'] = (
    session_df['time_diff'].isna()
    | (session_df['time_diff'] > pd.Timedelta(minutes=30))
)

session_df
```

```python
session_df['session_number'] = (
    session_df.groupby('customer_id')['new_session']
    .cumsum()
)

session_df.head(20)
```

```python
total_sessions = session_df['new_session'].sum()

print("Total estimated sessions:", total_sessions)
```

```python
session_df.groupby('customer_id')['session_number'].sum().reset_index(name= 'total_sessions').sort_values(by= 'total_sessions', ascending= False)
```

Most of the unregistered customers have browsed through the website. Other than that `customer_id`: `85, 67, 6`are the most active users

### Storage

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

load_dotenv(".env")

url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password=os.environ["MYSQL_PASSWORD"],
    host="localhost",
    port=3306,
    database="quickcart_db"
)

engine = create_engine(url)

print("SQLAlchemy engine created successfully!")
```

```python
# deleting rows if exist any:

from sqlalchemy import text

with engine.begin() as conn:

    # Delete child tables first
    conn.execute(text("DELETE FROM clickstream"))
    conn.execute(text("DELETE FROM orders"))

    # Delete parent tables
    conn.execute(text("DELETE FROM products"))
    conn.execute(text("DELETE FROM customers"))

print("Existing QuickCart records cleared!")
```

```python
customer_df.to_sql(
    name="customers",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
product_df.to_sql(
    name="products",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
order_df.to_sql(
    name="orders",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
valid_clickstream.to_sql(
    name="clickstream",
    con=engine,
    if_exists="append",
    index=False
)
```
