```sql
CREATE SCHEMA IF NOT EXISTS workspace.brewbeats;

SHOW SCHEMAS IN workspace;

USE SCHEMA brewbeats;

SHOW TABLES IN workspace.brewbeats;

DESC TABLE coffee_sales;

SELECT * FROM coffee_sales LIMIT 10;

-- Q1. Write a query to find the top 5 best-selling coffee types by revenue.
SELECT coffee_name, SUM(money)
FROM coffee_sales
GROUP BY 1
ORDER BY 2 DESC
LIMIT 3;

-- Q2. Write a query to find total revenue by payment type (cash_type).
SELECT cash_type, SUM(money) as total_revenue
FROM coffee_sales
WHERE cash_type ILIKE '%cash%'
GROUP BY 1;

-- Q3. Write a query using a CTE to find total revenue per month.
WITH revenue_per_month AS(
    SELECT MONTH(date) as monthly_sales, ROUND(SUM(money),2) as monthly_revenue
    FROM coffee_sales
    GROUP BY 1 
    ORDER BY 1
)

SELECT * FROM revenue_per_month;

-- Q4. Write a query to find the day with the highest number of transactions.

WITH coffees_sold as (
    SELECT DATE(date) as transaction_day, COUNT(*) as coffees_sold
    FROM coffee_sales
    GROUP BY 1
)
SELECT * , DENSE_RANK() OVER (ORDER BY coffees_sold DESC) AS DENSE_RANK
FROM coffees_sold
ORDER BY dense_rank ASC
LIMIT 1;


```
