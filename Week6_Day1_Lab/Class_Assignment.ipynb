Create a DataFrame with pyspark:

```python
data = [('Shahrukh','Khan','3/11/1965','M',3000),
  ('Salman','Khan','11/12/1967','M',4000),
  ('Hritik','Roshan','12/3/1969','M',4000),
  ('Alia','Bhatt','3/5/1993','F',4000),
  ('Anushka', 'Sharma','5/10/1989','F', 80000)
]

columns = ["firstname","lastname","dob","gender","salary"]
df = spark.createDataFrame(data=data, schema = columns)
df.printSchema()
df.show(3)
```

To read a CSV file from Volumes:

```python
df = spark.read.option("header",True) .csv("/Volumes/workspace/default/customers/customers.csv")
          
df.printSchema()
df.show(3)
```

createOrReplaceTempView: allows SQL to refer to that DataFrame.


```python
table_cust = spark.read.option("header",True) \
          .csv("/Volumes/workspace/default/customers/customers.csv") \
          .createOrReplaceTempView("customer")

spark.sql("SELECT customer_name, city FROM CUSTOMER").show(5)
```
