#### Q. 3 — Choose ETL or ELT

Choose the better starting pattern.



| Requirement                                                      | ETL or ELT?               | Reason                                   |
| ---------------------------------------------------------------- | --------------------------| ---------------------------------------- |
| Invalid data must be removed before it reaches the target        | ETL                       | Data must be transformed                 |
| Raw source data should be kept first for later reprocessing      | ELT                       | Data is first Extracted and Loaded       |
| Target system has strong SQL transformation capability           | ELT                       | Data is transformed at the target system |
| Sensitive fields must be removed before loading                  | ETL                       | Data is transformed before loading       |

- batch name

## Q. 7 — Preserve Raw Input Before Cleaning

```python
source = pd.DataFrame({
    "order_id": [301, 302, 303],
    "city": [" London ", None, "Leeds"],
    "amount_gbp": [45.50, 20.00, -5.00],
})
```

Create:

- raw_orders
- clean_orders

Rules for clean_orders:

- city must not be missing;
- amount must be positive;
- trim spaces from city.

Confirm that raw_orders still has all original rows.

```python
source.shape
```

```python
raw_orders= source.copy()
raw_orders.shape
```

```python
clean_orders= raw_orders.copy()
# Replacing null in city for Unknown:
clean_orders.loc[clean_orders['city'].isnull()== True, 'city']= 'Unknown'
```

```python
clean_orders
```

```python
# Amount must be positive:
clean_orders.drop(clean_orders[clean_orders['amount_gbp']<0].index, inplace= True)
clean_orders
```

```python
clean_orders['city']=clean_orders['city'].str.strip()
clean_orders
```

```python
print(f"Raw orders had {raw_orders.shape[0]} rows")
print(f"Clean orders had {clean_orders.shape[0]} rows")
```
