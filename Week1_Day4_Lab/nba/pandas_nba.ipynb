###  Data Cleaning 

```python
data.dtypes
```

##### Count how many missing (NaN) values are in each column. 

```python
data.isnull().sum()
```

##### Convert the Salary column to numeric (handle errors). 

##### Drop rows where Salary is missing. 

An error will be generated while converting the `Salary` column from float to int, because it contains NA values:



IntCastingNaNError: Cannot convert non-finite values (NA or inf) to integer: Error while type casting for column 'Salary'

```python
data['Salary'].isnull().sum()
```

```python
data[data['Salary'].isnull()]  #inspecting the rows with no salary
```

1. Drop the row where all the values are NULL.
2. Drop all the rows where `Salary` is NULL.
3. Convert `Salary` from float to int.
4. Replace NULL values to 'Unknown' in column `College`.

```python
data.dropna(subset= ['Salary'], inplace= True) #
```

```python
data['Salary'].isnull().sum()
```

```python
data= data.astype({'Salary': int})
data.info()
```

```python
data['College']= data['College'].fillna("Unknown")
```

```python
data.info()
```

###  Filtering Data

##### Show all players from Boston Celtics.

```python
data.loc[data['Team'] == 'Boston Celtics', ['Name']].reset_index(drop= True)
```

##### Show players whose Age is greater than 30. 

```python
data.loc[data['Age'] > 30, ['Name']].reset_index(drop= True)
```

##### Filter players who play as PG (Point Guard). 

```python
data[data['Position'] == 'PG']
```

##### Find players with Salary greater than 5,000,000.

```python
data[data['Salary'] > 5000000]
```

###  Sorting

##### Sort players by Salary (highest first). 

```python
data[data['Salary'] > 5000000].sort_values(by= 'Salary', ascending= False)
```

##### Sort players by Age (youngest first). 

```python
data.loc[:, ['Name', 'Team', 'Age']].sort_values(by= 'Age')
```

###    GroupBy Operations 

##### Find the average salary per team. 

```python
(data.groupby('Team')['Salary'].mean()).round(2)
```
