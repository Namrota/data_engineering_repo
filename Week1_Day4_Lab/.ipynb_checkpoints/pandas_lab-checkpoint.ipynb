```python
import pandas as pd
```

##### Series:

```python
#series is a labelled 1D array
s= pd.Series([10, 20, 30], name= 'amount')
print(s)
```

##### Dataframe:

```python
# Dataframe is a 2D tabular data with rows and columns:
df = pd.DataFrame({'amount':[10, 20, 30]})
print(df)
```

##### Q. 1 — Create a DataFrame

```python
data= [{'order_id': 101, 'city': 'London', 'amount': 45.50, 'status': 'complete'},
       {'order_id': 102, 'city': 'Manchester', 'amount': 18.80, 'status': 'cancelled'},
       {'order_id': 103, 'city': 'London', 'amount': 62.25, 'status': 'complete'},
       {'order_id': 104, 'city': 'Leeds', 'amount': 30.00, 'status': 'complete'}]
                    
order_df= pd.DataFrame(data)
order_df
```

```python
order_df.columns
```

```python
order_df.shape
```

##### Q.2 — Create and Read Three CSV Batches

```python
batch_1=pd.read_csv('case1.csv', sep= ';')
batch_2=pd.read_csv('case2.csv', sep= ';')
batch_3=pd.read_csv('case3.csv', sep= ';')

 
print("Batch 1 rows:", len(batch_1))
print("Batch 2 rows:", len(batch_2))
print("Batch 3 rows:", len(batch_3))
```

##### Q.3 — Combine the CSV Batches

```python
all_order_df= pd.concat([order_df, batch_1, batch_2, batch_3], ignore_index= True)
all_order_df
```
