#### Q. 14 — Revenue by City

Using only completed orders, calculate:

- total revenue by city;
- order count by city.
- Sort the result by revenue from highest to lowest.

```python
all_order_df.info()
```

```python
# total revenue by city: Series

all_order_df[all_order_df['status'] == 'complete'].groupby('city')['amount'].sum().reset_index(name= 'total_revenue')
```

Both produces same results, but the first returns a Series and the second returns a Dataframe. reset_index() works differently for both.

#### Q. 17 — Build a Three-Batch Cleaning Flow

```python
#Create three CSV batches
df1= pd.read_csv('batch_1.csv')
df2= pd.read_csv('batch_2.csv')
df3= pd.read_csv('batch_3.csv')

print("Three CSV batch files saved as a dataframe: \n")
print(df1)
print(df2)
print(df3)
```

Faced issue initially while concatenating:

**TypeError: cannot concatenate object of type '<class 'str'>'; only Series and DataFrame objs are valid**

Because I have used quotation marks for the dataframe ['df1', 'df2', 'df3'].


```python
# Combine the batches
combined_orders= pd.concat([df1, df2, df3], ignore_index= True)
combined_orders
```

Create:
- valid_orders;
- rejected_orders.
  
Reject:

- non-positive amount;
- missing city;
- duplicate order_id

```python
invalid_mask = ((combined_orders['city'].isna())|
    (combined_orders['amount_gbp'] < 0)|
    (combined_orders.duplicated(subset= 'order_id', keep= 'first'))
               )

rejected_orders= combined_orders.loc[invalid_mask].copy()

valid_orders= combined_orders.loc[~invalid_mask].copy()

print("Subset of rejected orders: ")
print(rejected_orders)
print("\nSubset of accepted orders:")
print(valid_orders)
```

```python
import numpy as np

duplicated_mask= combined_orders.duplicated(subset= 'order_id', keep= 'first')
conditions= [rejected_orders['city'].isna(),
            rejected_orders['amount_gbp'] < 0,
            duplicated_mask.loc[rejected_orders.index]
            ]

reasons= ['missing_city', 'non-positive amount', 'duplicate order_id']

rejected_orders['rejection_reason']= np.select(conditions, reasons, default= 'Unknown')

rejected_orders
```

```python
source_count= len(combined_orders)
valid_rows= len(valid_orders)
rejected_rows= len(rejected_orders)

print(f"Source Rows: {source_count}")
print(f"\nValid Rows: {valid_rows}")
print(f"\nInavlid Rows: {rejected_rows}")
```
