```python
import pandas as pd
```

```python
batch_1= pd.DataFrame([
(101, "London", 45.50, "complete"),
(102, "Manchester", -5.00, "complete"),
],columns=['order_id', 'City', 'Amount', 'Status'])
batch_2= pd.DataFrame([
(103, " Bristol ", 30.00, "complete"),
(104, "Leeds", 20.00, "cancelled"),
],columns=['order_id', 'City', 'Amount', 'Status'])
batch_3=pd.DataFrame([
(101, "London", 45.50, "complete"),
(105, "Manchester", 51.00, "complete"),
],columns=['order_id', 'City', 'Amount', 'Status'])

print(batch_1)
print(batch_2)
print(batch_3)
```

```python
bronze_df= pd.concat([batch_1, batch_2, batch_3], ignore_index=True)
print(bronze_df)
```

```python
bronze_df.shape
```

```python
# removing duplicates
silver_df= bronze_df.copy()
silver_df.drop_duplicates(subset= ['order_id'], keep= 'first', inplace= True)
silver_df
```

```python
silver_df.shape
```

```python
silver_df= silver_df[silver_df['Amount'] > 0].astype({'Amount':int})
silver_df
```

```python
silver_df.shape
```

```python
silver_df['City']= silver_df['City'].str.strip()
```

```python
silver_df.shape
```

After data validation now the dataset has 4 rows and 4 columns as compared to the raw data which had 6 rows and 4 columns.
