#### 2. Write a program that asks the user for two numbers and displays their addition, subtraction, multiplication, and division.

```python

def calculator(num1, num2, choice:int):
    if choice==1:
        return num1/num2
    elif choice==2:
        return num1 * num2
    elif choice==3:
        return num1 + num2
    else:
        return num1 - num2


# Dictionary for calculator
operations= {1: 'division', 2: 'multiplication', 3: 'addition', 4: 'subtraction'}

#user-input validation:
while True:
    user_choice= int(input(f"Enter your choice:\n1. Division \n2. Multiplication \n3. Addition \n4. Subtraction \n "))
    if user_choice> 4:
        print("Invalid Choice")
    else:
        num1= float(input("Enter your first number: "))
        num2= float(input("Enter your second number: "))
        result= calculator(num1, num2, user_choice)
        print(f"User chose {operations[user_choice]} for the numbers {num1} and {num2}. The result is {result}")
        user_inp= input("Do you want to continue [y/n]: ")
        if user_inp.lower()=='y':
            continue
        else: break
    

```

#### 24. Write a function count_greater(numbers, value) that accepts a list of numbers and a value, and returns how many numbers in the list are greater than that value.

```python
import random
def count_greater(numbers: list, value: int):
    count= 0
    max_lis= []
    for num in numbers:
        if num > value:
            count+=1
            max_lis.append(num)
    return count, max_lis
lis_of_items= [i for i in range(1,60)]
value= random.randrange(20, 80)
result, max_lis= count_greater(lis_of_items, value)
print(f"List of items: {lis_of_items}")
print(f"Random value generated: {value}")
if result== 0:
    print(f"No numbers greater than {value} is found.")
else:
    print(f"{result} numbers are greater than {value}: {max_lis}")
    
```
